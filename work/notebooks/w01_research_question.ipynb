## 2. The question: decision, action, cost of a wrong call



### Research question

Which pages should a content or SEO reviewer look at first for possible refresh or other intervention, based on observable evidence of decline, visibility, traffic, engagement, freshness, and opportunity?

**Decision:** Which pages should receive limited review capacity first?

**Action:** A reviewer investigates a high-ranked page and decides whether to refresh, expand, protect, prune, monitor, or take no action.

**Cost of a wrong recommendation:** A false positive can waste editorial or SEO review time on a page that did not need attention. A false negative can cause an important page that needed attention to be missed while review capacity is spent elsewhere.
